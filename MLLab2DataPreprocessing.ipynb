```python
#necessary library for handling missing values
from sklearn.impute import SimpleImputer

pima_df = pd.read_csv("/content/pima-indians-diabetes.csv")


#Identify missing values
#imputation is the name for replac the null
missing_data = pima_df.isnull().sum()
print("Missing values in each column:")
print(missing_data)

#Handling missing values
impute = SimpleImputer(missing_values=np.nan, strategy='mean')

impute.fit(pima_df)

df_imputed = impute.transform(pima_df)

#print (df_imputed)
#converting matrix into dataframe
df1 = pd.DataFrame(df_imputed, columns=pima_df.columns)
print (df1)
print("\nMissing values after handling:")
print(df1.isnull().sum())
print(df1)
```
